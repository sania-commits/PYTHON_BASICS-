```python
# Recursion: It's a technique in which a function call it-self directly.
# Iterative Vs Recursive (a*b):
# Iterative function for multiplication:

def multiply(a,b):
    result = 0
    for i in range(b):
        result = result + a
    print(result) 
multiply(3,4)       

# Recursive function for multiplication:

def mul(a,b):
    
    if b == 1: #base condition
        return a
    else:
        return a + mul(a,b-1)
    
print(mul(3,4))    
```

```python
# Iteration:
def factorial(number):
    fact = 1
    for i in range(1,number+1):
        fact = i * fact
    print(fact)
    
factorial(3)  

# Recursion:
def fact(number):
    if number == 1:
        return 1
    else:
        return number* fact(number-1)
print(fact(5))    
       
        
```

```python
# Palindrome: 
# Recursion:
def palin(text):
    if len(text) <= 1:
        print("Palindrome")
    else:
        if text[0] == text[-1]:
            palin(text[1:-1])
            print("Palindrome")
        else:
            print("Not a palindrome") 
            
print(palin("madam"))
print(palin("python"))                
        
```

```python
# Iteration:
def palin(text):
    n = len(text)
    for i in range( n//2 ):
        if text[i] == text[n-1-i]:
            return "palindrom"
        else:
            return " Not a Palindrome"
       
print(palin("madam"))        
        
```

```python
# The rabbit problem:
import time
def fib(m):
    if m == 0 or m == 1:
        return 1
    else:
        return fib(m-1) + fib(m-2)
start = time.time()
print(fib(12))
print(time.time() - start)    
```

```python
# Memoization:
import time
def memo(m,d):
    if m in d:
        return d[m]
    else:
        d[m] = memo(m-1,d) + memo(m-2,d)
        return d[m]

start = time.time()
d = {0:1, 1:1} 
print(memo(12 ,d))
print(time.time()-start)    
```

```python
# Power set:
def power_set(list):
    result =[]
    
    def helper(index,current):
        if index == len(list):
            result.append(current[:])
            return
        helper(index+1,current) # don't pick current element
        current.append(list[index])
        helper(index+1, current)
        
        current.pop() # backtrack
        
    helper(0,[])
    return result

nums = [1,2,3]
print(power_set(nums))        
```

```python
# Higher order function(HOF):
#L = [11,14,21,23,56,78,45,29,28]
#1. Print the sum of evens
#2. Print the sum of odds 
#3. Print the sum of numbers divisible by 3

# Normal function for this logic:
def return_sum(L):
    
    even_sum = 0
    odd_sum = 0
    div3_sum = 0
    
    for i in L:
        if i%2 == 0:
             even_sum = even_sum + i
        if i%2 != 0:
             odd_sum = odd_sum + i       
        if i%3 == 0:
            div3_sum = div3_sum + i  
        
    return(even_sum,odd_sum,div3_sum)
    
L = [11,14,21,23,56,78,45,29,28]
return_sum(L)

                   
```

```python
# Higher Order Function:
def return_sum(func, L):
    
    result = 0
    
    for i in L:
        if func(i):
            result = result + i
    
    return result        
    
L = [11,14,21,23,56,78,45,29,28]

x = lambda x: x%2 == 0 
y = lambda x: x%2 != 0 
z = lambda x: x%3 == 0

print(return_sum(x, L))
print(return_sum(y, L))
print(return_sum(z, L))
# the benifir of using higher order function:
# generally if you give input to a normal function then it work
# on it in a single way but you can give two things to a higher order function
# one is input and you can also tell the function how to work on that input
# In this previous eg you not even given input to the function but you
# also drived different behavior from this function using three lamba functions.
# And this is where lamba functions are used extensively.
```

```python
# Making our own for loop:
def OurOwnForLoop(iterable):
    
    iterator = iter(iterable) #fetching the iterator using iter function
    
    while True:
        
        try:
            print(next(iterator)) # calling next function
        except StopIteration:
            break       
```

```python
# Creating objects for OurOwnForLoop:
a = [1,2,3] #List
b = range(1,11) # Range
c = (1,2,3) # Tuple
d = {1,2,3} # Set
e = {0:1,1:1} # Dictionary

print(OurOwnForLoop(a))
print(OurOwnForLoop(b))
print(OurOwnForLoop(c))
print(OurOwnForLoop(d))
print(OurOwnForLoop(e))
```

```python
# Let's create our own range() function:
# iterable:
class my_range:
    
    def __init__(self,start,end):
        
        self.start = start
        self.end = end
        
    def __iter__(self):
        return my_range_iterator(self)

# iterator:            
class my_range_iterator:
    
    def __init__(self,iterable_obj):
        self.iterable = iterable_obj
        
    def __iter__(self):
        return self
    
    def __next__(self):
         
        if self.iterable.start >= self.iterable.end:
            raise StopIteration
        
        current = self.iterable.start
        self.iterable.start += 1
        return current
    
# creating object of my_range:
for i in my_range(1,11):
    print(i)           
```

```python
def square(num):
    for i in range(1,num+1):
        yield i**2
        
gen2 = square(10)

print(next(gen2))
print(next(gen2)) 
print(next(gen2)) 

for i in gen2:
    print(i)        
```

```python
# 1) User will input (3ages).Find the oldest one
age1 = int(input("Enter the first age"))
age2 = int(input("Enter the second age"))
age3 = int(input("Enter the third age")) 

def Oldest_Age():
    if age1 > age2 and age1 > age3 :
        return age1
    elif age2 > age1 and age2 > age3:
        return age2
    else:
        return age3  

oldest_age = Oldest_Age()    
print("Oldest age is:", oldest_age)         
        
```

```python
# 2) Write a program that will convert celsius value to fahrenheit
c = int(input("Enter the celsius temp"))

def Fahrenheit():
    return ((9*c)/5)+32

celsius = Fahrenheit()
print("celsius value to fahrenheit is", celsius)
    
```

```python
# 5) Write a program that will reverse a four digit number.Also it checks whether the reverse is true.
no = int(input("Enter the four digit no"))
number = no
reverse = 0
for i in range(0,4):
    digit = number%10
    reverse = reverse*10 +digit
    number = number // 10
   
    
print("The reverse four digit number is", reverse)
    
if no == reverse:
    print("True - Number is a palindrome")
else:
    print("False - Number is not a palindrome")    
```

```python
# 7)Write a program that will tell whether the given year is a leap year or not.
year = int(input("Enter a year: "))

if (year % 400 == 0) or (year % 4 == 0 and year % 100 != 0):
    print("Leap Year")
else:
    print("Not a Leap Year")
```

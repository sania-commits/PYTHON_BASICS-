```python
# Constructor
# Magic methods/ dunder methods
class Fraction:
    def __init__(self,n,d):
        self.num = n
        self.den = d
    
    def __str__(self):
        return "{}/{}".format(self.num, self.den)

    def __add__(self,other):
        temp_num = self.num * other.den + self.den *other.num
        temp_den = self.den*other.den
        return "{}/{}".format(temp_num, temp_den)
    
    def __sub__(self,other):
        temp_num = self.num * other.den - self.den *other.num
        temp_den = self.den*other.den
        return "{}/{}".format(temp_num, temp_den)
    
    def __mul__(self,other):
        temp_num = self.num * other.num
        temp_den = self.den*other.den
        return "{}/{}".format(temp_num, temp_den)
    
    def __truediv__(self,other):
        temp_num = self.num * other.den
        temp_den = self.den * other.num
        return "{}/{}".format(temp_num, temp_den)
    
x = Fraction(3,4)
y = Fraction(5,6)

print(x+y)
print(x-y)
print(x*y)
print(x/y)
    
    
                
        
```

```python
# Pass by reference: when we pass the object reference to the function then it is called pass by reference.
class Customer:
    
    def __init__(self,name):
        self.name = name
    
def greet(customer): #  here we are taking the object reference as an argument in the function greet.
    print("Hello", customer.name)
    print(id(customer))
    
cust = Customer("Zaid")# here cust is the object reference or a reference variable. 
print(id(cust))
greet(cust)   # here we are passing the object reference cust to the function greet.  

# So the id's of cust and customer will be the same because both are pointing to the same object in the memory.       
```

```python
# Static method: Static method is a method which is defined inside the class but it does not take self as a parameter.
# it is used to perform some operation which is not related to the object of the class.
# it is defined using @staticmethod decorator.
class Customer:
    
    def __init__(self, name, age):
        self.name = name
        self.age = age
        
    @staticmethod
    def greet():
        print("Hello, welcome to our store!")
c1 = Customer("Zaid", 25)
c1.greet() # here we are calling the static method greet using the object reference c1. 
#but it is not necessary to call the static method using the object reference. 
# we can also call the static method using the class name. for eg: Customer.greet() 
# it will also work because static method is not related to the object of the class. it is related to the class itself. 
Customer.greet() # here we are calling the static method greet using the class name Customer
```

```python
# Static Counter: Static counter is a variable which is defined inside the class and it is shared by all the objects of the class.
# it is used to count the number of objects created for a class.
class Customer:
    
    counter = 0 # this is a static counter which is shared by all the objects of the class. 
    #it is used to count the number of objects created for a class.
    
    def __init__(self,name,age):
        self.name = name
        self.age = age
        Customer.counter += 1 # here we are incrementing the static counter by 1 every time an object is created for the class Customer.
        
c1 = Customer("Zaid", 25)
print(Customer.counter) # here we are accessing the static counter using the class name Customer. it will print 1 because only one object is created for the class Customer.
c2 = Customer("Sania", 24)
print(Customer.counter) # here we are accessing the static counter using the class name Customer. it will print 2 because two objects are created for the class Customer.
c3 = Customer("Ali", 26)
print(Customer.counter) # here we are accessing the static counter using the class name Customer. it will print 3 because three objects are created for the class Customer. 
    
     
         
```

```python
# Relationship between two classes: when two classes are related to each other then it is called relationship between two classes.
# There are three types of relationship between two classes: 1) Association 2) Aggregation 3) Composition

# Association: when two classes are related to each other but they are not dependent on each other then it is called association.
# For eg: class Car and class Engine are related to each other but they are not dependent on each other because a car can have an 
# engine and it can also run without an engine.

# Aggregation: when two classes are related to each other and they are dependent on each other then it is called aggregation.
# For eg: class Car and class Engine are related to each other and they are dependent on each other because a car
# cannot run without an engine and an engine cannot run without a car.

# Composition: when two classes are related to each other and they are dependent on each other and one class cannot exist without the other class then it is called composition.
# For eg: class Car and class Engine are related to each other and they are dependent on each other and one class cannot exist without the other class 
# because a car cannot exist without an engine and an engine cannot exist without a car.

# BUT MAINLY IN PYTHON WE USE ASSOCIATION AND AGGREGATION. COMPOSITION IS NOT USED MUCH IN PYTHON BECAUSE IT IS NOT A STRICTLY OOP LANGUAGE.
# In python we can achieve composition using aggregation but in other languages like java, c++ etc we cannot achieve composition using aggregation because they are strictly OOP languages.
```

```python
# Aggregation:
class Customer:
    
    def __init__(self,name,gender,address):
        self.name = name
        self.gender = gender
        self.address = address
        
    def edit_profile(self , new_name , new_city, new_pincode, new_state): 
        self.name = new_name
        self.address.change_address(new_city, new_pincode, new_state) # here we are calling the change_address method of the Address class using the address object reference. 
        return "Profile updated successfully" # here we performed aggregation because the Customer class is using the Address class to perform some operation. Achived "Has a" relationship between Customer and Address class.
        
class Address:
    
    def __init__(self,city,pincode,state):
        self.city = city
        self.pincode = pincode
        self.state = state
        
    def change_address(self, new_city, new_pincode, new_state):
        self.city = new_city
        self.pincode = new_pincode
        self.state = new_state
        return "Address updated successfully"
        
add = Address("Delhi", 110001, "Delhi")
cust = Customer("Zaid","Male", add)

cust.edit_profile(new_name = "Zaid Malik", new_city = "Noida", new_pincode = 201301, new_state = "Uttar Pradesh") # here we are changing the city, pincode and state of the customer using the edit_profile method of the Customer class.
print(cust.address)
print(cust.address.city)            
```

```python
# Polymorphism: poly means many and morphism means phases.
# so polymorphism means "Multiple Phases"
class Phone:
    def __init__(self, price, brand, camera):
        print("Inside phone constructor")
        self.__price = price # added __ in price and make it private class
        self.brand = brand
        self.camera = camera
            
    def buy(self): # here parent and child class have same method buy.
        print("Buying a phone")
            
    def return_phone(self):
        print("Returning a phone")
            
class FeaturePhone(Phone):
    pass

class SmartPhone(Phone):
    def buy(self):   # if the child class and parent class have the same method then the child class executes its owm method not of parent class.
        print("Buying a smartphone")

s = SmartPhone(20000,"Apple",13)
print(s.buy())
# this is called method overriding which comes under polymorphism.
```

```python
# 2) Method Overloading: 

# you can induce different behaviour by giving
# different inputs to the same method.
# Technically method overloading doesn't exist in python.
class Geometry:
    
    def area(self,a):
        return 3.14*a*a
    
    def area(self,l,b):
        return l*b

obj = Geometry()
print(obj.area(4))    
print(obj.area(4,5))
```

```python
# By appling some tricks to impliment method overloading:
class Geometry:
    
    def area(self,a,b=0):
        if b==0:
            print("Circle", 3.14*a*a)
        else:
            print("Rectangle", a*b)
            
obj = Geometry()
obj.area(4)
obj.area(4,5)                
```

```python
# Example of Supar:
class Phone:
    def __init__(self, price, brand, camera):
        print("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera
        
    def buy(self):
        print("Buying a phone")
        
    def return_phone(self):
        print("Returning a phone")
        
class FeaturePhone(Phone):
    pass
class SmartPhone(Phone):
    def buy(self): # it will be executed due to method overriding
        print("Buying a smartphone") 
        super().buy() # You can invoke the parent's constructor and methods but not attributes throught super keyword.
        
s = SmartPhone(20000,"Apple",13)
print(s.buy())   
# but super keyword can't work outside the class                  
```

```python
# Super example with Constructor:
class Phone:
    def __init__(self,price,brand,camera): # follows DRY-(don't repeate your code) principal
        print("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera
        
    def buy(self):
        print("Buying a phone")
        
    def return_phone(self):
        print("Returning a phone")
        
class FeaturePhone(Phone):
    pass
class SmartPhone(Phone):
    def __init__(self,price,brand,camera,os,ram):
        print("Inside smartphone constructor first time")
        super().__init__(price,brand,camera) # here super with const. invoke the parent class const. for not to repeat the code for same variables.
        self.os = os
        self.ram = ram
        print("Inside smartphone constructor")
        
    def buy(self):
        print("Buying a smartphone")
        
s = SmartPhone(20000,"Samsung",12,"Android",2)
print(s.os)  
print(s.brand)          
              
```

```python
# Example on super:
class Parent:
    def __init__(self,num):
        self.__num = num
    def get_num(self):
        return self.__num

class Child(Parent):
    def __init__(self,num,val):
        super().__init__(num) # This super call should be a first statement inside the constructor.
        self.__val = val
    
    def get_val(self):
        return self.__val
    
son = Child(100,200)
print(son.get_num())
print(son.get_val())                
```

```python
# Example 2 on super:
class Parent:
    def __init__(self):
        self.num = 100

class Child(Parent):
    def __init__(self):
        super().__init__()
        self.var = 200
    
    def show(self):
        print(self.num) # we can access self.num inside the class because self is nothing but an object son. i.e self= son
        # so if son can access this self.num then self can also do so .
        print(self.var)
        
son = Child()
print(son.show())                    
```

```python
# Example on super:
class Parent:
    def __init__(self):
        self.__num = 100
        
    def show(self):
        print("Parent:", self.__num)
        
class Child(Parent):
    def __init__(self):
        super().__init__()
        self.__var = 10
        
    def show(self):
        print("Child:",self.__var)
        
dad = Parent()
print(dad.show())

son = Child()
print(son.show())                       
```

```python
# Types of inheritance:
# 1) Single level 2)Multi level 3) Hierarchical 4) Multiple 5) Hybrid
```

```python
# 1) Single level inheritance:
class Phone:
    def __init__(self,price,brand,camera):
        print("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera
        
    def buy(self):
        print("Buying a phone")
        
    def return_phone(self):
        print("Returning a phone")
        
class SmartPhone(Phone): #here SmartPhone class inherit the Phone class
    pass

SmartPhone(1000,"Apple","13px").buy()                
```

```python
# 2) Multi-level inheritance:
class Product:
    def review(self):
        print("Product customer review")
        
class Phone(Product):
    def __init__(self,price,brand,camera):
        print("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera
    def buy(self):
        print("Buying a phone")
        
    def return_phone(self):
        print("Returning a phone")
        
class SmartPhone(Phone): # here SmartPhone inherit the Phone class 
                         #and phone class inherit the Product class.
    pass

s = SmartPhone(20000,"Apple",12)
print(s.buy())
print(s.review())                
```

```python
# 3) Hierarchical inheritance:
class Phone: # Single Parent
    def __init__(self,price,brand,camera):
        print("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera
    def buy(self):
        print("Buying a phone")
        
    def return_phone(self):
        print("Returning a phone")
        
class SmartPhone(Phone): # Child1
    pass
class FeaturePhone(Phone):# Child2
    pass

SmartPhone(1000,"Apple","13px").buy()
```

```python
# 4) Multiple inheritance:
class Phone: # Parent class1
    def __init__(self,price,brand,camera):
        print("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera
    def buy(self):
        print("Buying a phone")
        
    def return_phone(self):
        print("Returning a phone")
        
class Product: #Parent class2
    def review(self):
        print("Customer review")        
        
class SmartPhone(Phone, Product): 
    # here This child class inherit two parent classes.
    pass

s = SmartPhone(2000,"Apple", 12)
print(s.buy())
print(s.review())
```

```python
# MRO( Method Resolution Order) Example:
class Phone:
    def __init__(self,price,brand,camera):
        print("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera
    
    def buy(self):
        print("Buying a phone")
        
    def return_phone(self):
        print("Returning a phone")
        
class Product:
    def buy(self): # this buy method is executed by child(smartphone) class because
        print("Product buy method")
        
class SmartPhone(Product, Phone): # you have written Product class at first 
    # This is called MRO(Method Resolution Order)
    pass

s = SmartPhone(2000,"Apple",12)  
print(s.buy())                     
```

```python
# Example 1 on Types:
class A: # Dada
    def m1(self):
        return 20

class B(A): # Papa
    def m1(self): # this will be used by child class
        return 30
    def m2(self):
        return 40

class C(B): # Child
    def m2(self):
        return 20
    
obj1 = A()
obj2 = B()
obj3 = C() # here child will use his father's method rather than using his dada's m1 method.
print(obj1.m1()+obj3.m1()+obj3.m2()) # 20+30+20 = 70           
```

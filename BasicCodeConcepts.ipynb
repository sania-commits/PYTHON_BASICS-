```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n*factorial(n-1)
print(factorial(5))    
```

```python
def factorial(n):
    while n > 1:
        return n*factorial(n-1)
    return 1
print(factorial(1))
```

```python
def factorial(n):
    fact = 1 
    while n > 0:
        fact *= n
        n -= 1
    return fact
print(factorial(5))
    
```

```python
def factorial(n):
    fact = 1
    for i in range(1, n+1):
        fact *= i
    return fact

n = 5
print(f"Factorial of {n} is {factorial(n)}")
```

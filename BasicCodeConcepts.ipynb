```python
#operators:
#arithmetic operators: +, -, *, /, //, %, **
#comparison operators: ==, !=, >, <, >=, <=
#logical operators: and, or, not
#assignment operators: =, +=, -=, *=, /=, //=, %=,
#bitwise operators: &, |, ^, ~, <<, >>
#identifiers: is, is not
#membership operators: in, not in

#1. Arithmetic operators:
x = 5
y = 2
print(x + y)  # addition
print(x - y)  # subtraction
print(x * y)  # multiplication
print(x / y)  # True division
print(x // y)  # floor division(integer division)
print(x % y)  # modulus
print(x ** y)  # exponentiation

#2. Comparison operators:
a = 10
b = 20
print(a == b)  # equal to
print(a != b)  # not equal to
print(a > b)  # greater than
print(a < b)  # less than
print(a >= b)  # greater than or equal to
print(a <= b)  # less than or equal to

#3. Logical operators:
p = True
q = False
print(p and q)  # logical AND
print(p or q)  # logical OR
print(not p)  # logical NOT

#4. Bitwise operators:
m = 5  # in binary: 0101
n = 3  # in binary: 0011
print(m & n)  # bitwise AND (0101 & 0011 = 0001) => 1 (0 if any bit is 0, otherwise 1)
print(m | n)  # bitwise OR (0101 | 0011 = 0111) => 7 (1 if any bit is 1, otherwise 0)
print(m ^ n)  # bitwise XOR (0101 ^ 0011 = 0110) => 6 (1 if the bits are different, otherwise 0)
print(~m)  # bitwise NOT (~0101 = 1010) => -6 (in two's complement representation) (returns the complement of the number, which is the number with all bits inverted)
print(m << 2)  # left shift (0101 << 2 = 10100) => 20 (shift the bits to the left by a specified number of positions, filling the rightmost bits with zeros)
print(m >> 1)  # right shift (0101 >> 1 = 0010) => 2 (shift the bits to the right by a specified number of positions, filling the leftmost bits with the sign bit for signed numbers or with zeros for unsigned numbers)

#5. Assignment operators:
d = 5
print(d)  # assignment
d += 3  # addition assignment
print(d)  # 8
d -= 2  # subtraction assignment
print(d)  # 6
d *= 2  # multiplication assignment
print(d)  # 12
d /= 2  # division assignment
print(d)  # 6.0
d //= 2  # floor division assignment
print(d)  # 3.0
d %= 2  # modulus assignment
print(d)  # 1.0
d **= 2  # exponentiation assignment
print(d)  # 1.0

i = 10
i &= 1  # bitwise AND assignment
print(i)  # 1.0
i |= 1  # bitwise OR assignment
print(i)  # 1.0
i ^= 1  # bitwise XOR assignment
print(i)  # 0.0

#6. Identity operators:
e = [1, 2, 3]
f = [1, 2, 3]
print(e is f)  # False (e and f are different objects in memory)
print(e is not f)  # True (e and f are different objects in memory)
g = e
print(e is g)  # True (e and g refer to the same object in memory)

#7. Membership operators:
h = [1, 2, 3, 4, 5]
print(3 in h)  # True (3 is an element of the list h)
print(6 in h)  # False (6 is not an element of the list h
print(3 not in h)  # False (3 is an element of the list h)
print(6 not in h)  # True (6 is not an element of the list h)
```

```python
#Break Statement:
for i in range(1, 11):
    if i == 5:
        break  # exit the loop when i is equal to 5
    print(i)  # prints numbers from 1 to 4
```

```python
#Example 1: Using break in a while loop
counter = 0
while True:
    print(counter)
    counter += 1
    if counter == 5:
        break  # exit the loop when counter is equal to 5   
```

```python
#Example: Break is used in Linear search algorithm to exit the loop 
# when the target element is found in the list.
arr = [1, 2, 3, 4, 5]
target = 3
for i in range(len(arr)):
    if arr[i] == target:
        print(f"Element found at index {i}")
        break  # exit the loop when the target element is found
else:
    print("Element not found in the list")  
```

```python
#continue statement:
# The continue statement is used to skip the current iteration of a loop and move on to the next iteration. 
# When the continue statement is executed, the rest of the code inside the loop for the current iteration is skipped, 
# and the loop proceeds to the next iteration.
for i in range(1, 11):
    if i % 2 != 0:
        continue  # skip the rest of the code inside the loop for odd numbers
    print(i)  # prints even numbers from 1 to 10
```

```python
#Strings:
#A string is a sequence of characters enclosed in single quotes, double quotes or triple quotes.
#Strings are immutable which means you cannot change the characters in a string after it is created.
#Strings can be indexed and sliced to access individual characters or substrings.
#String literals can be defined using single quotes, double quotes or triple quotes.
#1. Single quotes:
name = 'Sania'
print(name)
#2. Double quotes:
greeting = "Hello, World!"
print(greeting)
#3. Triple quotes:
multiline_string = """This is a multiline string."""
print(multiline_string)
#4. Raw string:
raw_string = r"C:\Users\Sania\Documents"
print(raw_string)
#5. Unicode string:
unicode_string = u"\U0001F600\U0001F606\U0001F923"  # grinning face emoji
print(unicode_string)

#creating a string using str() function:
number = 123
number_str = str(number)
print(number_str)  # Output: '123'
print(type(number_str))  # Output: <class 'str'>

#accessing characters in a string using indexing:
name = "Sania"
print(name[0])  # Output: 'S'
print(name[1])  # Output: 'a'
print(name[2])  # Output: 'n'
print(name[3])  # Output: 'i'
print(name[4])  # Output: 'a'

# Concept of Indexing and slicing in strings:
name = "Sania Anjum"
print(name[0])  # Output: 'S'
print(name[1])  # Output: 'a'
print(name[2])  # Output: 'n'
print(name[3])  # Output: 'i'
print(name[4])  # Output: 'a'
print(name[5])  # Output: ' ' (space character)
print(name[6])  # Output: 'A'
print(name[7])  # Output: 'n'
print(name[8])  # Output: 'j'
print(name[9])  # Output: 'u'
print(name[10])  # Output: 'm'

#Type of indexing:
#1. Positive indexing: starts from 0 and goes up to n-1 (where n is the length of the string)
#2. Negative indexing: starts from -1 and goes up to -n (where n is the length of the string)
name = "Sania Anjum"
print(name[-1])  # Output: 'm'
print(name[-2])  # Output: 'u'
print(name[-3])  # Output: 'j'
print(name[-4])  # Output: 'n'
print(name[-5])  # Output: 'A'
print(name[-6])  # Output: ' ' (space character)
print(name[-7])  # Output: 'a'
print(name[-8])  # Output: 'i'
print(name[-9])  # Output: 'n'
print(name[-10])  # Output: 'a'
print(name[-11])  # Output: 'S'

#concept of slicing in strings:
name = "Sania Anjum"
print(name[0:5])  # Output: 'Sania' (slices the string from index 0 to index 4)
print(name[6:11])  # Output: 'Anjum' (slices the string from index 6 to index 10)
print(name[:5])  # Output: 'Sania' (slices the string from the beginning to index 4)
print(name[6:])  # Output: 'Anjum' (slices the string from index 6 to the end)
print(name[:])  # Output: 'Sania Anjum' (slices the entire string)      
print(name[::2])  # Output: 'Sna nju' (slices the string with a step of 2, which means it takes every second character starting from index 0)
print(name[1::2])  # Output: 'aiAum' (slices the string with a step of 2, which means it takes every second character starting from index 1)
print(name[2:6:2])  # Output: 'n ' (slices the string from index 2 to index 5 with a step of 2, which means it takes every second character starting from index 2)

#negative slicing:
name = "Sania Anjum"
print(name[-11:-6])  # Output: 'Sania' (slices the string from index -11 to index -7)
print(name[-5:])  # Output: 'Anjum' (slices the string from index -5 to the end)
print(name[:-6])  # Output: 'Sania ' (slices the string from the beginning to index -7)
print(name[-6:-1])  # Output: 'Anju' (slices the string from index -6 to index -2)
print(name[-1:])  # Output: 'm' (slices the string from index -1 to the end)
print(name[-1:-6:-1])  # Output: 'mujnA' (slices the string from index -1 to index -5 with a step of -1, which means it takes every second character starting from index -1 and moving backwards)   
print(name[-6:-11:-1])  # Output: 'AniaS' (slices the string from index -6 to index -10 with a step of -1, which means it takes every second character starting from index -6 and moving backwards)
print(name[-1:-12:-1])  # Output: 'mujnA iaS' (slices the string from index -1 to index -11 with a step of -1, which means it takes every second character starting from index -1 and moving backwards) 
print(name[-1::-1])  # Output: 'mujnA iaS' (slices the string from index -1 to the beginning with a step of -1, which means it takes every second character starting from index -1 and moving backwards)
print(name[::-1])  # Output: 'mujnA iaS' (slices the entire string with a step of -1, which means it takes every second character starting from the end and moving backwards)
print(name[-1:-12:-2])  # Output: 'mjnAia' (slices the string from index -1 to index -11 with a step of -2, which means it takes every second character starting from index -1 and moving backwards)
print(name[-2::-2])  # Output: 'juAia' (slices the string from index -2 to the beginning with a step of -2, which means it takes every second character starting from index -2 and moving backwards)
print(name[::-1]) # Reverse the stringOutput: 'Sania Anjum' (slices the entire string with a step of 1, which means it takes every character starting from the beginning to the end)   

#adding characters to a string using concatenation:
first_name = "Sania"
last_name = "Anjum"
full_name = first_name + " " + last_name
print(full_name)  # Output: 'Sania Anjum'

#Escaping characters in a string using backslash (\):
quote = "She said, \"Hello!\""
print(quote)  # Output: 'She said, "Hello!"'

#editing strings: strings are immutable which means you cannot change the characters in a string after it is created. 
c = "ArithmeticError"
print(c)  # Output: 'ArithmeticError'
# c[0] = 'a'  # This will raise a TypeError because strings are immutable and do not support item assignment. 
# You cannot change the character at index 0 of the string 'c' to 'a'.
# However, you can create a new string by concatenating the existing string with the new characters.
city = "Kolkata"
print(city)  # Output: 'Kolkata'
city = city + " is a beautiful city"
print(city)  # Output: 'Kolkata is a beautiful city'
#You can also use string formatting to create a new string with the desired characters.
city = "Kolkata"
print(city)  # Output: 'Kolkata'
city = f"{city} is a beautiful city"
print(city)  # Output: 'Kolkata is a beautiful city'
#You can also use the replace() method to create a new string by replacing certain characters in the existing string with new characters.
city = "Kolkata"
print(city)  # Output: 'Kolkata'
city = city.replace("Kolkata", "Kolkata is a beautiful city")
print(city)  # Output: 'Kolkata is a beautiful city'
#You can also use the join() method to create a new string by joining a list of strings with a specified separator.
city = "Kolkata"
print(city)  # Output: 'Kolkata'
city = " ".join([city, "is a beautiful city"])
print(city)  # Output: 'Kolkata is a beautiful city'
#You can also use the format() method to create a new string by replacing placeholders in the existing string with new characters.
city = "Kolkata"
print(city)  # Output: 'Kolkata'
city = "{} is a beautiful city".format(city)
print(city)  # Output: 'Kolkata is a beautiful city'
#You can also use the % operator to create a new string by replacing placeholders in the existing string with new characters.
city = "Kolkata"
print(city)  # Output: 'Kolkata'
city = "%s is a beautiful city" % city
print(city)  # Output: 'Kolkata is a beautiful city'
#You can also use the f-string syntax to create a new string by replacing placeholders in the existing string with new characters.
city = "Kolkata"
print(city)  # Output: 'Kolkata'
city = f"{city} is a beautiful city"
print(city)  # Output: 'Kolkata is a beautiful city'
#You can also use the string concatenation operator (+) to create a new string by concatenating the existing string with the new characters.
city = "Kolkata"
print(city)  # Output: 'Kolkata'
city = city + " is a beautiful city"
print(city)  # Output: 'Kolkata is a beautiful city'
#You can also use the string multiplication operator (*) to create a new string by repeating the existing string a specified number of times.
city = "Kolkata"
print(city)  # Output: 'Kolkata'
city = city * 3
print(city)  # Output: 'KolkataKolkataKolkata'
#You can also use the string slicing syntax to create a new string by slicing the existing string and concatenating it with the new characters.
city = "Kolkata" 
print(city)  # Output: 'Kolkata'
city = city[:4] + " is a beautiful city"
print(city)  # Output: 'Kolk is a beautiful city'
#You can also use the string slicing syntax to create a new string by slicing the existing string and concatenating it with the new characters.
city = "Kolkata"
print(city)  # Output: 'Kolkata'
city = city[:4] + " is a beautiful city" + city[4:]
print(city)  # Output: 'Kolkata is a beautiful city' 
#you can also reassign a new value to the existing string variable to create a new string with the desired characters. 
name = "Sania"
print(name)  # Output: 'Sania'
name = "Sania Anjum"
print(name)  # Output: 'Sania Anjum'

#deleting a string:
name = "Sania"
print(name)  # Output: 'Sania'
del name
# print(name)  # This will raise a NameError because the variable 'name' has been deleted and is no longer defined.


#String methods:

#1. upper(): returns a new string with all characters converted to uppercase
name = "Sania"
print(name.upper())  # Output: 'SANIA'

#2. lower(): returns a new string with all characters converted to lowercase
name = "Sania"
print(name.lower())  # Output: 'sania'

#3. title(): returns a new string with the first character of each word capitalized
name = "sania anjum"
print(name.title())  # Output: 'Sania Anjum'

#4. strip(): returns a new string with leading and trailing whitespace removed
name = "   Sania   "
print(name.strip())  # Output: 'Sania'

#5. replace(): returns a new string with all occurrences of a substring replaced with another substring
greeting = "Hello, World!"
print(greeting.replace("World", "Sania"))  # Output: 'Hello, Sania!'

#6. split(): returns a list of substrings separated by a specified delimiter
sentence = "Hello, how are you?"
print(sentence.split())  # Output: ['Hello,', 'how', 'are', 'you?']

#7. join(): returns a new string that is the concatenation of the strings in a list, separated by a specified delimiter
words = ['Hello', 'World']
print(" ".join(words))  # Output: 'Hello World'

#8. find(): returns the index of the first occurrence of a substring in a string, or -1 if the substring is not found
greeting = "Hello, World!"
print(greeting.find("World"))  # Output: 7
print(greeting.find("Sania"))  # Output: -1

#9. count(): returns the number of occurrences of a substring in a string
sentence = "Hello, how are you? Hello!"
print(sentence.count("Hello"))  # Output: 2

#10. startswith(): returns True if a string starts with a specified substring, otherwise returns False
greeting = "Hello, World!"
print(greeting.startswith("Hello"))  # Output: True
print(greeting.startswith("World"))  # Output: False

#11. endswith(): returns True if a string ends with a specified substring, otherwise returns False
greeting = "Hello, World!"
print(greeting.endswith("!"))  # Output: True
print(greeting.endswith("Hello"))  # Output: False

#12. isalpha(): returns True if all characters in a string are alphabetic, otherwise returns False
name = "Sania"
print(name.isalpha())  # Output: True
name_with_space = "Sania Anjum"
print(name_with_space.isalpha())  # Output: False (because of the space character)

#13. isdigit(): returns True if all characters in a string are digits, otherwise returns
False
number_str = "12345"
print(number_str.isdigit())  # Output: True
number_str_with_space = "123 45"
print(number_str_with_space.isdigit())  # Output: False (because of the space character

#14. isalnum(): returns True if all characters in a string are alphanumeric (letters and digits), otherwise returns False
username = "Sania123"
print(username.isalnum())  # Output: True
username_with_space = "Sania 123"
print(username_with_space.isalnum())  # Output: False (because of the space character)
```

```python
#Operations on strings:
#1. Arithmetic operations:

# a. Concatenation: the + operator can be used to concatenate two or more strings
first_name = "Sania"
last_name = "Anjum"
full_name = first_name + " " + last_name    
print(full_name)  # Output: 'Sania Anjum'

# b. Repetition: the * operator can be used to repeat a string a specified number of times
greeting = "Hello! "
print(greeting * 3)  # Output: 'Hello! Hello! Hello! '  

#2. Comparison operations:

# a. Equality: the == operator can be used to compare two strings for equality
name1 = "Sania"
name2 = "Sania"
name3 = "Anjum"
print(name1 == name2)  # Output: True
print(name1 == name3)  # Output: False

# b. Inequality: the != operator can be used to compare two strings for inequality
name1 = "Sania"
name2 = "Sania"
name3 = "Anjum"
print(name1 != name2)  # Output: False
print(name1 != name3)  # Output: True

# c. Lexicographical comparison: the <, >, <=, >= operators can be used to compare two strings 
# lexicographically (based on the Unicode code points of the characters)
name1 = "Sania"
name2 = "Anjum"
print(name1 < name2)  # Output: False (because 'S' has a higher Unicode code point than 'A')
print(name1 > name2)  # Output: True (because 'S' has a higher Unicode code point than 'A')
print(name1 <= name2)  # Output: False (because 'S' has a higher Unicode code point than '  A')

#3. Logical operations:
# a. Membership: the in operator can be used to check if a substring is present in a string
sentence = "Hello, how are you?"
print("Hello" in sentence)  # Output: True
print("Sania" in sentence)  # Output: False
print("how" in sentence)  # Output: True
print("hi" in sentence)  # Output: False
# b. Identity: the is operator can be used to check if two string variables refer to the same string object in memory
name1 = "Sania"
name2 = "Sania"
name3 = name1
print(name1 is name2)  # Output: False (because name1 and name2 are different string objects in memory)
print(name1 is name3)  # Output: True (because name1 and name3 refer to the same string object in memory)   
# c. Logical operators: the and, or, not operators can be used to combine multiple string comparisons
name1 = "Sania"
name2 = "Anjum"
name3 = "Sania Anjum"
print(name1 == "Sania" and name2 == "Anjum")  # Output: True
print(name1 == "Sania" or name2 == "Anjum")  # Output: True
print(not name1 == "Anjum")  # Output: True 

print("" and "Sania")  # Output: '' (because the empty string is considered False in a logical context)
print("Sania" or "")  # Output: 'Sania' (because the empty string is considered False in a logical context)
print(not "")  # Output: True (because the empty string is considered False in a logical context)

#4. Bitwise operations: bitwise operations are not applicable to strings in python, as strings are not treated as sequences of bits.
# However, you can perform bitwise operations on the Unicode code points of the characters 
# in a string by using the ord() function to get the code point of each character and then
# applying the bitwise operators to those code points.
name = "Sania"
for char in name:
    char_code = ord(char)  # Get the Unicode code point of the character
    print(f"Character: {char}, Code point: {char_code}, Bitwise NOT: {~char_code}, Left shift by 2: {char_code << 2}, Right shift by 1: {char_code >> 1}")
    
#5. Assignment operations: assignment operations can be used to assign a new value to a string variable, 
# but they do not modify the original string object since strings are immutable.    
name = "Sania"
print(name)  # Output: 'Sania'
name = "Sania Anjum"
print(name)  # Output: 'Sania Anjum'    

#6. Identity operations: identity operations can be used to check if two string variables refer to the same string object in memory,
# but they do not modify the original string objects since strings are immutable.
name1 = "Sania"
name2 = "Sania"
name3 = name1
print(name1 is name2)  # Output: False (because name1 and name2 are different string objects in memory)
print(name1 is name3)  # Output: True (because name1 and name3 refer to the same string object in memory) 

#7. Membership operations: membership operations can be used to check if a substring is present in a string,
# but they do not modify the original string since strings are immutable.
sentence = "Hello, how are you?"
print("Hello" in sentence)  # Output: True
print("Sania" in sentence)  # Output: False
print("how" in sentence)  # Output: True
print("hi" in sentence)  # Output: False 

#8.loops on strings: you can use loops to iterate over the characters in a string, 
# but they do not modify the original string since strings are immutable.
name = "Sania"
for char in name[0:5:2]:  # iterating over the string with a step of 2, which means it takes every second character starting from index 0
    print(char)  # Output: 'S', 'a', 'n', 'i', 'a' (each character is printed on a new line)    
    
```

```python
#Common fuctions used with strings:
#1. len(): returns the length of a string
name = "Sania Anjum"
print(len(name))  # Output: 11 (including the space character)
#2. type(): returns the type of a variable
name = "Sania"
print(type(name))  # Output: <class 'str'>
#3. str(): used to convert a value to a string
number = 123
number_str = str(number)
print(number_str)  # Output: '123'
#5. sorted(): returns a sorted list of the characters in a string
name = "Sania"
print(sorted(name))  # Output: ['A', 'a', 'a', 'i', 'n', 'S'] (the characters are sorted in alphabetical order) 
#6. reversed(): returns an iterator that yields the characters in a string in reverse order
name = "Sania"
print(list(reversed(name)))  # Output: ['a', 'i', 'n',  'a', 'S'] (the characters are returned in reverse order)    
#7. enumerate(): returns an iterator that yields pairs of index and character for each character in a string
name = "Sania"
for index, char in enumerate(name):
    print(f"Index: {index}, Character: {char}")  # Output: Index: 0, Character: 'S' (and so on for each character in the string)    
#8. zip(): can be used to combine two or more strings into a single iterable of tuples, where each tuple contains one character from each string
string1 = "Sania"
string2 = "Anjum"
for char1, char2 in zip(string1, string2):
    print(f"String 1 character: {char1}, String 2 character: {char2}")  # Output: String 1 character: 'S', String 2 character: 'A' (and so on for each pair of characters from the two strings) 
#9. any(): returns True if any character in a string is true (i.e., not an empty string or a string containing only whitespace), otherwise returns False
name = "Sania"
print(any(name))  # Output: True (because the string is not empty and does not contain only whitespace)
empty_string = ""
print(any(empty_string))  # Output: False (because the string is empty)
whitespace_string = "   "
print(any(whitespace_string))  # Output: False (because the string contains only whitespace)
#10. all(): returns True if all characters in a string are true (i.e., not an empty string or a string containing only whitespace), otherwise returns False
name = "Sania"
print(all(name))  # Output: True (because all characters in the string are not empty and do not contain only whitespace)
empty_string = ""
print(all(empty_string))  # Output: False (because the string is empty)
whitespace_string = "   "
print(all(whitespace_string))  # Output: False (because the string contains only whitespace)  
#11. ord(): returns the Unicode code point of a character
char = 'A'
print(ord(char))  # Output: 65 (the Unicode code point for 'A')
char = 'a'
print(ord(char))  # Output: 97 (the Unicode code point for 'a')     
#12. chr(): returns the character corresponding to a Unicode code point
unicode_code = 65
print(chr(unicode_code))  # Output: 'A'
unicode_code = 97
print(chr(unicode_code))  # Output: 'a'
#13. format(): used to format a string by replacing placeholders with specified values
name = "Sania"
age = 26
message = "My name is {} and I am {} years old.".format(name, age)
print(message)  # Output: My name is Sania and I am 26 years old.
#14. f-string: used to format a string by embedding expressions inside curly braces {} within a string literal prefixed with 'f' or 'F'
name = "Sania"
age = 26        
message = f"My name is {name} and I am {age} years old."
print(message)  # Output: My name is Sania and I am 26 years old
#15. isspace(): returns True if all characters in a string are whitespace, otherwise returns False
whitespace_string = "   "
print(whitespace_string.isspace())  # Output: True (because all characters in the string are whitespace)
non_whitespace_string = "Sania"
print(non_whitespace_string.isspace())  # Output: False (because not all characters in the string are whitespace)
mixed_string = "Sania Anjum"
print(mixed_string.isspace())  # Output: False (because not all characters in the string are whitespace)            
#16. islower(): returns True if all characters in a string are lowercase, otherwise returns False
lowercase_string = "hello"
print(lowercase_string.islower())  # Output: True (because all characters in the string are lowercase)
mixed_string = "Hello"          
print(mixed_string.islower())  # Output: False (because not all characters in the string are lowercase)
uppercase_string = "HELLO"
print(uppercase_string.islower())  # Output: False (because not all characters in the string are lowercase)
#17. isupper(): returns True if all characters in a string are uppercase, otherwise returns False
uppercase_string = "HELLO"
print(uppercase_string.isupper())  # Output: True (because all characters in the string are uppercase)
mixed_string = "Hello"
print(mixed_string.isupper())  # Output: False (because not all characters in the string are uppercase)
lowercase_string = "hello"
print(lowercase_string.isupper())  # Output: False (because not all characters in the   string are uppercase)   
#18. istitle(): returns True if the string is in title case (i.e., the first character of each word is uppercase and the rest are lowercase), otherwise returns False
title_string = "Hello World"
print(title_string.istitle())  # Output: True (because the string is in title case)
```

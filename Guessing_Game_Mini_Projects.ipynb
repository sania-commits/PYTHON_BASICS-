```python
#Create Custom function instead of title() method:
def custom_title(s):
    words = s.split()  # Split the string into words
    title_cased_words = [word.capitalize() for word in words]  # Capitalize the first letter of each word
    return " ".join(title_cased_words)  # Join the capitalized words back into a single string
input_string = "hello world"
output_string = custom_title(input_string)
print(output_string)  # Output: "Hello World" (the input string is converted to title case using the custom_title function)
input_string = "sania anjum is a student"
output_string = custom_title(input_string)
print(output_string)  # Output: "Sania Anjum Is A Student" (the input string is converted to title case using the custom_title function)
input_string = "python programming is great"
output_string = custom_title(input_string)
print(output_string)  # Output: "Python Programming Is Great" (the input string is converted to title case using the custom_title function)
```

```python
#Create custom fuction and take input fron user and check if the string is in title case or not:
def is_title_case(s):
    words = s.split()  # Split the string into words
    for word in words:
        if not word[0].isupper() or not word[1:].islower():  # Check if the first letter is uppercase and the rest are lowercase
            return False  # If any word does not meet the title case criteria, return False
    return True  # If all words meet the title case criteria, return True
input_string = input("Enter a string: ")
if is_title_case(input_string):
    print("The string is in title case.")
else:    print("The string is not in title case.")      
```

```python
#Create a custom fuction to fetch the email address of a user by splitting it from @ and taking input from user:
def get_email_address():
    email = input("Enter your email address: ")  # Take input from the user
    if "@" in email:  # Check if the email address contains the "@" symbol
        username, domain = email.split("@")  # Split the email address into username and domain
        return username  # Return the username
    else:
        return "Invalid email address"  # Return an error message if the input is not a valid email address
email_address = get_email_address()
print("Your username is:", email_address)  # Output: Your username is: "The username extracted from the email address" (the username extracted from the email address is returned and printed)
     

```

```python
#Create a function to remove dublicates from a list taking input from user:
def remove_duplicates():
    input_list = input("Enter a list of elements separated by commas: ").split(",")  # Take input from the user and split it into a list
    unique_elements = list(set(input_list))  # Remove duplicates by converting the list to a set and back to a list
    return unique_elements  # Return the list of unique elements
unique_list = remove_duplicates()
print("List with duplicates removed:", unique_list)  # Output: List with duplicates removed: "The list of unique elements entered by the user" (the list of unique elements entered by the user is returned and printed)  
print(sorted(unique_list))  # Output: "The list of unique elements sorted in ascending order" (the list of unique elements is sorted in ascending order and printed)
```

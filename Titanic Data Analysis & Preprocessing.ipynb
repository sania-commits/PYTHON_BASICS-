LOAD DATASET:

```python
import pandas as pd

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"

df = pd.read_csv(url)

print(df.head(10))
print(df.describe())
```

```python
print(df.info())
print(df.isnull().sum())
```

```python
print(df.columns)
print(df.describe())
```

UNIVARIATE ANALYSIS ON AGE COLUMN:


**FILLING MISSING VALUES:**

```python
df['Age'] = df['Age'].fillna(df['Age'].mean())
```

**CONVERT TO INT:**

```python
df['Age'] = df['Age'].astype(int)
```

```python
print(df.info())
```

```python
print(df.isnull().sum())
```

```python
df['Age'].describe()
```

```python
df['Age'].plot(kind='hist', bins=20)
```

Most of the passengers are under the age of 30years. 

```python
df['Age'].plot(kind='kde')
```

The curve is normally distributed as the value of skewness is almost equal to 1.

```python
df['Age'].plot(kind='box')
```
